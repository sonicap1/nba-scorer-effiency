```python
scoring_df = df[
    [
        "PLAYER_NAME",
        "TEAM_ABBREVIATION",
        "GP",
        "MIN",
        "PTS",
        "FGM",
        "FGA",
        "FG_PCT",
        "FG3M",
        "FG3A",
        "FG3_PCT",
        "FTM",
        "FTA",
        "FT_PCT"
    ]
].copy()

scoring_df.head()
```

```python
scoring_df["PTS_PER_GAME"] = scoring_df["PTS"] / scoring_df["GP"]

scoring_df[
    ["PLAYER_NAME", "GP", "PTS", "PTS_PER_GAME"]
].head()
```

```python
scoring_df[["GP", "MIN"]].describe()
```

```python
(scoring_df["MIN"] >= 500).sum()
```

```python
(scoring_df["MIN"] >= 1000).sum()
```

```python
qualified_df = scoring_df[scoring_df["MIN"] >= 500].copy()

qualified_df.shape
```

```python
qualified_df["TS_PCT"] = qualified_df["PTS"] / ( 
    2 * (qualified_df["FGA"] + 0.44 * qualified_df["FTA"])
)

qualified_df[
    ["PLAYER_NAME", "PTS_PER_GAME", "FG_PCT", "TS_PCT"]
].head()
```

```python
qualified_df["TS_PCT"].describe()
```

```python
(qualified_df["PTS_PER_GAME"] >= 10).sum()
```

```python
(qualified_df["PTS_PER_GAME"] >= 15).sum()
```

```python
(qualified_df["PTS_PER_GAME"] >= 20).sum()
```

```python
scorers_df = qualified_df[
    qualified_df["PTS_PER_GAME"] >= 15
].copy()

scorers_df.shape
```

```python
scorers_df[
    ["PLAYER_NAME", "TEAM_ABBREVIATION", "PTS_PER_GAME", "TS_PCT"]
].sort_values(
    by="TS_PCT",
    ascending=False
).head(10)
```

```python
scorers_df[
    ["PLAYER_NAME", "TEAM_ABBREVIATION", "PTS_PER_GAME", "TS_PCT"]
].sort_values(
    by="PTS_PER_GAME",
    ascending=False
).head(10)
```

```python
import plotly.express as px

fig = px.scatter(
    scorers_df,
    x="PTS_PER_GAME",
    y="TS_PCT",
    hover_name="PLAYER_NAME",
    title="Scoring Volume vs. Scoring Efficiency"
)

fig.show()
```

```python
correlation = scorers_df["PTS_PER_GAME"].corr(
    scorers_df["TS_PCT"]
)

correlation
```

```python
scorers_df["THREE_POINT_ATTEMPT_RATE"] = (
    scorers_df["FG3A"] / scorers_df["FGA"]
)

scorers_df[
    ["PLAYER_NAME", "FG3A", "FGA", "THREE_POINT_ATTEMPT_RATE"]
].head(10)
```

```python
scorers_df["FREE_THROW_RATE"] = (
    scorers_df["FTA"] / scorers_df["FGA"]
)

scorers_df[
    ["PLAYER_NAME", "FTA", "FGA", "FREE_THROW_RATE"]
].head(10)
```

```python
three_point_correlation = scorers_df["THREE_POINT_ATTEMPT_RATE"].corr(
    scorers_df["TS_PCT"]
)

free_throw_correlation = scorers_df["FREE_THROW_RATE"].corr(
    scorers_df["TS_PCT"]
)

three_point_correlation, free_throw_correlation
```

```python
three_point_pct_correlation = scorers_df["FG3_PCT"].corr(
    scorers_df["TS_PCT"]
)

three_point_pct_correlation
```

```python
free_throw_pct_correlation = scorers_df["FT_PCT"].corr(
    scorers_df["TS_PCT"]
)

free_throw_pct_correlation
```

```python
fg_pct_correlation = scorers_df["FG_PCT"].corr(
    scorers_df["TS_PCT"]
)

fg_pct_correlation
```

```python
scorers_df["FG2_PCT"] = (
    (scorers_df["FGM"] - scorers_df["FG3M"])
    / (scorers_df["FGA"] - scorers_df["FG3A"])
)

scorers_df[
    ["PLAYER_NAME", "FG_PCT", "FG2_PCT", "FG3_PCT", "TS_PCT"]
].head(10)
```

```python
two_point_pct_correlation = scorers_df["FG2_PCT"].corr(
    scorers_df["TS_PCT"]
)

two_point_pct_correlation
```

```python
import statsmodels.api as sm
```

```python
X = scorers_df[
    [
        "FG2_PCT",
        "FG3_PCT",
        "FT_PCT",
        "THREE_POINT_ATTEMPT_RATE",
        "FREE_THROW_RATE",
        "PTS_PER_GAME"
    ]
]

y = scorers_df["TS_PCT"]
```

```python
X = sm.add_constant(X)

model = sm.OLS(y, X).fit()

print(model.summary())
```

```python
import pandas as pd
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

vif_data = pd.DataFrame()

vif_data["Variable"] = X.columns

vif_data["VIF"] = [
    variance_inflation_factor(X.values, i)
    for i in range(X.shape[1])
]

vif_data
```
