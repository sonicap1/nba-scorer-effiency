```python
scoring_df = df[
    [
        "PLAYER_NAME",
        "TEAM_ABBREVIATION",
        "GP",
        "MIN",
        "PTS",
        "FGM",
        "FGA",
        "FG_PCT",
        "FG3M",
        "FG3A",
        "FG3_PCT",
        "FTM",
        "FTA",
        "FT_PCT"
    ]
].copy()

scoring_df.head()
```

```python
scoring_df["PTS_PER_GAME"] = scoring_df["PTS"] / scoring_df["GP"]

scoring_df[
    ["PLAYER_NAME", "GP", "PTS", "PTS_PER_GAME"]
].head()
```

```python
scoring_df[["GP", "MIN"]].describe()
```

```python
(scoring_df["MIN"] >= 500).sum()
```

```python
(scoring_df["MIN"] >= 1000).sum()
```

```python
qualified_df = scoring_df[scoring_df["MIN"] >= 500].copy()

qualified_df.shape
```

```python
qualified_df["TS_PCT"] = qualified_df["PTS"] / ( 
    2 * (qualified_df["FGA"] + 0.44 * qualified_df["FTA"])
)

qualified_df[
    ["PLAYER_NAME", "PTS_PER_GAME", "FG_PCT", "TS_PCT"]
].head()
```

```python
qualified_df["TS_PCT"].describe()
```

```python
(qualified_df["PTS_PER_GAME"] >= 20).sum()
```

```python
scorers_df = qualified_df[
    qualified_df["PTS_PER_GAME"] >= 15
].copy()

scorers_df.shape
```

```python
scorers_df[
    ["PLAYER_NAME", "TEAM_ABBREVIATION", "PTS_PER_GAME", "TS_PCT"]
].sort_values(
    by="TS_PCT",
    ascending=False
).head(10)
```

```python
import statsmodels.api as sm
```

```python
X = scorers_df[
    [
        "FG2_PCT",
        "FG3_PCT",
        "FT_PCT",
        "THREE_POINT_ATTEMPT_RATE",
        "FREE_THROW_RATE",
        "PTS_PER_GAME"
    ]
]

y = scorers_df["TS_PCT"]
```

```python
X = sm.add_constant(X)

model = sm.OLS(y, X).fit()

print(model.summary())
```

```python
import pandas as pd
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

vif_data = pd.DataFrame()

vif_data["Variable"] = X.columns

vif_data["VIF"] = [
    variance_inflation_factor(X.values, i)
    for i in range(X.shape[1])
]

vif_data
```

```python
from sklearn.preprocessing import StandardScaler

predictor_columns = [
    "FG2_PCT",
    "FG3_PCT",
    "FT_PCT",
    "THREE_POINT_ATTEMPT_RATE",
    "FREE_THROW_RATE",
    "PTS_PER_GAME"
]

scaler = StandardScaler()

X_standardized = scaler.fit_transform(
    scorers_df[predictor_columns]
)
```

```python
X_standardized = pd.DataFrame(
    X_standardized,
    columns=predictor_columns
)

X_standardized.head()
```

```python
y_standardized = (
    scorers_df["TS_PCT"] - scorers_df["TS_PCT"].mean()
) / scorers_df["TS_PCT"].std()
```

```python
X_standardized = sm.add_constant(X_standardized)

standardized_model = sm.OLS(
    y_standardized.reset_index(drop=True),
    X_standardized
).fit()

print(standardized_model.summary())
```

```python
top_efficient_scorers = scorers_df[
    [
        "PLAYER_NAME",
        "TEAM_ABBREVIATION",
        "PTS_PER_GAME",
        "TS_PCT",
        "FG2_PCT",
        "FG3_PCT",
        "FT_PCT",
        "THREE_POINT_ATTEMPT_RATE",
        "FREE_THROW_RATE"
    ]
].sort_values(
    by="TS_PCT",
    ascending=False
).head(10)

top_efficient_scorers
```

```python
profiles_columns = [
    "FG2_PCT",
    "FG3_PCT",
    "FT_PCT",
    "FREE_THROW_RATE"
]

scorers_df[profiles_columns].mean()
```

```python
profile_means = scorers_df[profiles_columns].mean()

profile_differences = scorers_df[
    ["PLAYER_NAME"] + profiles_columns
].copy()

for column in profiles_columns:
    profile_differences[column] = (
        profile_differences[column] - profile_means[column]
    )

profile_differences.head()
```

```python
top_scorer_profiles = top_efficient_scorers[
    ["PLAYER_NAME", "TS_PCT"] + profiles_columns
].copy()

for column in profiles_columns:
    top_scorer_profiles[column] = (
        top_scorer_profiles[column] - profile_means[column]
    )

top_scorer_profiles
```

```python
player_name = "Shai Gilgeous-Alexander"

player_profile = profile_differences[
    profile_differences["PLAYER_NAME"] == player_name
].copy()

player_profile
```

```python
player_profile_long = player_profile.melt(
    id_vars="PLAYER_NAME",
    value_vars=profiles_columns,
    var_name="Metric",
    value_name="Difference_From_Average"
)

player_profile_long
```

```python
fig = px.bar(
    player_profile_long,
    x="Metric",
    y="Difference_From_Average",
    title=f"{player_name} Scoring Profile vs. Average Volume Scorer"
)

fig.add_hline(y=0)

fig.show()
```

```python
profile_means = scorers_df[profiles_columns].mean()
profile_stds = scorers_df[profiles_columns].std()

profile_zscores = scorers_df[
    ["PLAYER_NAME"] + profiles_columns
].copy()

for column in profiles_columns:
    profile_zscores[column] = (
        profile_zscores[column] - profile_means[column]
    ) / profile_stds[column]

profile_zscores.head()
```

```python
player_name = "Shai Gilgeous-Alexander"

player_zprofile = profile_zscores[
    profile_zscores["PLAYER_NAME"] == player_name
].copy()

player_zprofile
```

```python
player_zprofile_long = player_zprofile.melt(
    id_vars="PLAYER_NAME",
    value_vars=profiles_columns,
    var_name="Metric",
    value_name="Z_Score"
)

player_zprofile_long
```

```python
fig = px.bar(
    player_zprofile_long,
    x="Metric",
    y="Z_Score",
    title=f"{player_name} Scoring Profile vs. Average Volume Scorer"
)

fig.add_hline(y=0)

fig.show()
```

```python
players_to_compare = [
    "Shai Gilgeous-Alexander",
    "Jalen Duren",
    "Jimmy Butler III"
]

comparison_profiles = profile_zscores[
    profile_zscores["PLAYER_NAME"].isin(players_to_compare)
].copy()

comparison_profiles
```

```python
comparison_long = comparison_profiles.melt(
    id_vars="PLAYER_NAME",
    value_vars=profiles_columns,
    var_name="Metric",
    value_name="Z_Score"
)

comparison_long
```

```python
fig = px.bar(
    comparison_long,
    x="Metric",
    y="Z_Score",
    color="PLAYER_NAME",
    barmode="group",
    title="Scoring Profiles of Efficient NBA Scorers"
)

fig.add_hline(y=0)

fig.show()
```
